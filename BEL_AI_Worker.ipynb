```
import nest_asyncio
nest_asyncio.apply()

import os
import cv2
import json
import uuid
import uvicorn
from fastapi import FastAPI, UploadFile, File, Form
from fastapi.middleware.cors import CORSMiddleware
from ultralytics import YOLO
import easyocr
from pyngrok import ngrok

# Initialize Models
print("Loading YOLO11...")
v_model = YOLO("yolo11n.pt")  # Vehicle tracker
print("Loading EasyOCR...")
ocr = easyocr.Reader(['en'], gpu=True)

app = FastAPI()
app.add_middleware(CORSMiddleware, allow_origins=["*"], allow_methods=["*"], allow_headers=["*"])

os.makedirs("/content/videos", exist_ok=True)

@app.post("/process")
async def process_video(video: UploadFile = File(...), camera_id: str = Form(...)):
    file_id = str(uuid.uuid4())
    vid_path = f"/content/videos/{file_id}.mp4"
    
    with open(vid_path, "wb") as f:
        f.write(await video.read())
        
    print(f"[{camera_id}] Video received. Processing on T4 GPU...")
    
    cap = cv2.VideoCapture(vid_path)
    total_frames = int(cap.get(cv2.CAP_PROP_FRAME_COUNT))
    frame_skip = 3
    frame_count = 0
    best_plates = {}
    
    while cap.isOpened():
        ret, frame = cap.read()
        if not ret: break
        frame_count += 1
        if frame_count % frame_skip != 0: continue
        
        r = v_model.track(frame, classes=[2,3,5,7], persist=True, verbose=False, tracker="bytetrack.yaml")[0]
        if r.boxes.id is not None:
            track_ids = r.boxes.id.int().cpu().tolist()
            boxes = r.boxes.xyxy.cpu().numpy()
            for t_id, box in zip(track_ids, boxes):
                x1, y1, x2, y2 = map(int, box)
                vehicle_crop = frame[max(0, y1):y2, max(0, x1):x2]
                if vehicle_crop.size == 0: continue
                
                res = ocr.readtext(vehicle_crop)
                for line in res:
                    text, conf = line[1], line[2]
                    text = text.replace(" ", "").upper()
                    if len(text) >= 5 and conf > 0.4:
                        if t_id not in best_plates or conf > best_plates[t_id]['conf']:
                            best_plates[t_id] = {'plate': text, 'conf': float(conf), 'frame': frame_count}
                                
    cap.release()
    os.remove(vid_path)
    
    results = []
    for v_id, data in best_plates.items():
        results.append({"plate_number": data['plate'], "confidence": data['conf']})
        
    print(f"[{camera_id}] Processing complete. Extracted {len(results)} plates.")
    return {"status": "success", "camera_id": camera_id, "results": results}

port = 8000
public_url = ngrok.connect(port).public_url
print("========================================================")
print("✅ COLAB GPU WORKER IS ONLINE!")
print(f"👉 Copy this URL into your local dashboard: {public_url}")
print("========================================================")

uvicorn.run(app, host="0.0.0.0", port=port)
```
